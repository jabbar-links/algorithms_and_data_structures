## Problem Two: Path to Leaves
**Description**  
Given the `root` of a binary tree, return all root to leaf paths in any order.

**Examples**

* Example 1:

    ![](./images/q1_ex1.png)

    Input: `root = [1, 2, 2, 3, 5, 6, 7]` *What traversal method is this?*

    Output: [[1, 2, 3], [1, 2, 5], [1, 2, 6], [1, 2, 7]]

* Example 2:

    ![](./images/q1_ex3.png)

    Input: `root = [10, 9, 7, 8]`

    Output: [[10, 7], [10, 9, 8]]

**Problem 2 Starter Code**

You may assume that `root` is a binary tree that has been built, i.e., you can access a node/leaf by calling `root.left.right.left.right...`, so you don't have to provide a tree-building function to convert an input list to a binary tree as an object of the `TreeNode` class.

```python
# Definition for a binary tree node.
class TreeNode(object):
     def __init__(self, val = 0, left = None, right = None):
         self.val = val
         self.left = left
         self.right = right
```

```python
from typing import List

def bt_path(root: TreeNode) -> List[List[int]]:
    paths = []
    current_path = []

    def dfs(node):
        if node is None:
            return

        current_path.append(node.val)

        if node.left is None and node.right is None:
            paths.append(current_path.copy())
        else:
            dfs(node.left)
            dfs(node.right)

        current_path.pop()  # Inside dfs: remove after visiting this node.

    dfs(root)  # Outside dfs: start exploring the tree.
    return paths
```

```python
# Example 1
root1 = TreeNode(
    1,
    left=TreeNode(2, left=TreeNode(3), right=TreeNode(5)),
    right=TreeNode(2, left=TreeNode(6), right=TreeNode(7))
)

print("Example 1:", bt_path(root1))

# Example 2
root2 = TreeNode(
    10,
    left=TreeNode(9, left=TreeNode(8)),
    right=TreeNode(7)
)

print("Example 2:", bt_path(root2))
```


4. Copy the solution your partner wrote. 


```python
# His code  here

from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:

    #Sort the list using in-house sort method for lists

    nums.sort()

    #Determine the length of the list

    length = len(nums)

    #now we iterate and find the zeros. use while loop
    #We need to track the items checked and we stop when the original list finishes

    items_checked = 0
    i = 0  #index

    while items_checked < length:
        # Check if the items equals 0
        if nums[i] == 0:
            # temporarily remove it from the list
            nums.pop(i) #pop the 0
            nums.append(0) #Add it to the end of the list
            # We dont need to increase the index here because the next items shifted left into index i when we popped
        else:
            i += 1 #in this case move to the next item if it wasn't 0.

        items_checked += 1 #Keep track of the progress through the original list

    return nums
```
